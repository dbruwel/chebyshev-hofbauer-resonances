```python
# imports
import matplotlib.pyplot as plt
import numpy as np

from chebyshev_hofbauer_resonances.general_tent_map.approx_transfer_op import (
    approx_super_adjacency,
)
```

```python
# function definition
def t_map(a):
    function_domains = [
        (0, 0.5),
        (0.5, 1),
        (1, 1.5),
        (1.5, 2),
    ]
    functions = [
        lambda x: a * x,
        lambda x: a * (1 - x),
        lambda x: 2 - a * (x - 1),
        lambda x: 2 - a * (2 - x),
    ]
    inverses = [
        lambda y: y / a,
        lambda y: 1 - y / a,
        lambda y: (2 - y) / a + 1,
        lambda y: (y - 2) / a + 2,
    ]
    derivatives = [
        lambda _: a,
        lambda _: -a,
        lambda _: -a,
        lambda _: a,
    ]

    return function_domains, functions, inverses, derivatives
```

```python
lam2s = []

for alpha in np.linspace(1.1, 2, 901):
    function_domains, functions, inverses, derivatives = t_map(alpha)

    super_adjacency = approx_super_adjacency(
        function_domains,
        functions,
        inverses,
        derivatives,
        N=5,
        K=5,
        depth=50,
    )

    evals_super_adj = np.linalg.eigvals(super_adjacency)
    evals_super_adj = evals_super_adj[np.argsort(-np.abs(evals_super_adj))]

    lam2 = evals_super_adj[1]
    lam2s.append(lam2)
```
