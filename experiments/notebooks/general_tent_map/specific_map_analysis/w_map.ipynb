```python
# imports
import matplotlib.pyplot as plt
import numpy as np

from chebyshev_hofbauer_resonances.general_tent_map.approx_transfer_op import (
    approx_super_adjacency,
)
```

```python
# function definition
a = 1.2
function_domains = [
    (0, 0.25),
    (0.25, 0.5),
    (0.5, 0.75),
    (0.75, 1),
]
functions = [
    lambda x: 1 - 4 * x,
    lambda x: (2 + a) * (x - 0.25),
    lambda x: 0.5 + 0.25 * a - (2 + a) * (x - 0.5),
    lambda x: 4 * (x - 0.75),
]
inverses = [
    lambda y: (1 - y) / 4,
    lambda y: y / (2 + a) + 0.25,
    lambda y: -(y - 0.5 - 0.25 * a) / (2 + a) + 0.5,
    lambda y: y / 4 + 0.75,
]
derivatives = [
    lambda _: -4,
    lambda _: 2 + a,
    lambda _: -2 + a,
    lambda _: 4,
]
```

```python
super_adjacency = approx_super_adjacency(
    function_domains,
    functions,
    inverses,
    derivatives,
    N=10,
    K=10,
    depth=100,
)

evals_super_adj = np.linalg.eigvals(super_adjacency)
evals_super_adj = evals_super_adj[np.argsort(-np.abs(evals_super_adj))]
```
